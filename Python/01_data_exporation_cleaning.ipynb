# Data Exploration

```python
import pandas as pd
import os

DATA_PATH = "../data/raw"

for file in os.listdir(DATA_PATH):
    if file.endswith(".csv"):
        print(file)
```

```python
customers = pd.read_csv(f"{DATA_PATH}/customers.csv")
orders = pd.read_csv(f"{DATA_PATH}/orders.csv")
order_items = pd.read_csv(f"{DATA_PATH}/order_items.csv")
products = pd.read_csv(f"{DATA_PATH}/products.csv")
payments = pd.read_csv(f"{DATA_PATH}/payments.csv")
returns = pd.read_csv(f"{DATA_PATH}/returns.csv")
marketing = pd.read_csv(f"{DATA_PATH}/marketing.csv")
```

```python
datasets = {
    "customers": customers,
    "orders": orders,
    "order_items": order_items,
    "products": products,
    "payments": payments,
    "returns": returns,
    "marketing": marketing
}

for name, df in datasets.items():
    print(f"{name:15} → {df.shape}")
```

### Missing Values and data types fix

```python
for name, df in datasets.items():
    print(f"\n{name.upper()}")
    
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    
    if len(missing):
        print("Missing values:")
        print(missing)
    else:
        print("No missing values")
```

## Data Integrity Checks


### Check Foreign keys

```python
# A. Every order should have a valid customer

orders[~orders["customer_id"].isin(customers["customer_id"])]
```

```python
orders["customer_id"].isin(customers["customer_id"]).value_counts()
```

```python
# B. Every order item should belong to an existing order

order_items[
    ~order_items["order_id"].isin(orders["order_id"])
]
```

```python
# C. Every order item should reference an existing product

order_items[
    ~order_items["product_id"].isin(products["product_id"])
]
```

```python
# D. Every payment should reference an existing order

returns[
    ~returns["order_id"].isin(orders["order_id"])
]
```

```python
# E. Every return should reference an existing order

returns[
    ~returns["order_id"].isin(orders["order_id"])
]
```

### Validate Quantities

```python
order_items["quantity"].describe()
```

```python
order_items[
    (order_items["quantity"] <= 0) |
    (order_items["quantity"].isna())
]
```

### Validate Prices

```python
products[
    (products["unit_price"] <= 0) |
    (products["unit_cost"] <= 0)
]
```

```python
order_items[
    (order_items["unit_price"] <= 0) |
    (order_items["unit_cost"] <= 0)
]
```

### Validate discount percentages

```python
order_items["discount_pct"].describe()
```

```python
order_items[
    (order_items["discount_pct"] < 0) |
    (order_items["discount_pct"] > 100)
]
```

### Validate Calculations

```python
# Gross Amount

calculated_gross = (
    order_items["quantity"] *
    order_items["unit_price"]
).round(2)

(order_items["gross_amount"] != calculated_gross).sum()
```

```python
# Discount 

calculated_discount = (
    order_items["gross_amount"] *
    order_items["discount_pct"] / 100
).round(2)

(order_items["discount_amount"] != calculated_discount).sum()
```

```python
# Net Amount

calculated_net = (
    order_items["gross_amount"] -
    order_items["discount_amount"]
).round(2)

(order_items["net_amount"] != calculated_net).sum()
```
